```python
import os 
from dotenv import load_dotenv
import psycopg2

load_dotenv()

try:
    con = psycopg2.connect(
        host=os.getenv("DB_HOST"),
        port=os.getenv("DB_PORT"),
        database=os.getenv("DB_NAME"),
        user=os.getenv("DB_USER"),
        password=os.getenv("DB_PASSWORD"),
    )
except Exception as erro:
    print(f"Ocorreu um erro: {erro}")

#comitar transações
con.autocommit = True

# objeto para executar consultas no banco de dados
cur = con.cursor()
cur.execute('CREATE DATABASE inventario;')
con.close()
```

```python
import os 
from dotenv import load_dotenv
import psycopg2

load_dotenv()

try:
    con = psycopg2.connect(
        host=os.getenv("DB_HOST"),
        port=os.getenv("DB_PORT"),
        database="inventario",
        user=os.getenv("DB_USER"),
        password=os.getenv("DB_PASSWORD"),
    )
except Exception as erro:
    print(f"Ocorreu um erro: {erro}")

#comitar transações
con.autocommit = True

# objeto para executar consultas no banco de dados
cur = con.cursor()
cur.execute('CREATE TABLE arquivos (idArquivo INT, nomeArquivo VARCHAR(256));')
con.close()
```

```python
import boto3
import io 
import psycopg2
import os 
from dotenv import load_dotenv

load_dotenv()

s3 = boto3.resource(
    service_name='s3',
    region_name=os.getenv('region_nameS3'),
    aws_access_key_id=os.getenv('aws_access_key_id'),
    aws_secret_access_key=os.getenv('aws_secret_access_key')
)

bucket = os.getenv('bucket_name')
prefixo = 'imagens/'

for objetos_s3 in s3.Bucket(bucket).objects.filter(Prefix=prefixo):
    if objetos_s3.key.endswith('jpg') or objetos_s3.key.endswith('JPG'):
        filename = objetos_s3.key.split('/')[1]
        print(filename)
```
